Plot customization

```python
x=np.array([2023,2024,2025,2026])
y1=np.array([15,25,30,20])
y2=np.array([17,23,38,5])
y3=np.array([13,15,20,30])
line_style=dict(marker="o",
         markersize=10,
         markerfacecolor="blue",
         markeredgecolor="blue",
         linestyle="solid",
         linewidth=2,)
plt.plot(x,y1,color="black",**line_style)
plt.plot(x,y2,color="red",**line_style)
plt.plot(x,y3,color="green",**line_style)
plt.show()
```

Labels

```python
x=np.array([2023,2024,2025,2026])
y1=np.array([15,25,30,20])
y2=np.array([17,23,38,5])
y3=np.array([13,15,20,30])
plt.title("Class size",
          fontsize=20,
          family="Arial",
          fontweight="bold",
          color="#041c26")
plt.xlabel("Year 2025",fontsize=20,
           family="Arial",
           fontweight="bold",
           color="#041c26")
plt.ylabel("Students",fontsize=20,
           family="Arial",
           fontweight="bold",
           color="#041c26")
plt.tick_params(axis="both",
                colors="#02212e")
plt.xticks(x)
plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3)
plt.show()
```

Bar Chart

```python
#bar chart -> compare categories of data
categories=["Grains","Fruit","Vegetable","Protein","Diary","Sweets"]
values=np.array([4,3,2,5,3,1])
#plt.grid(axis="y")
#plt.bar(categories,values,color="skyblue")
#horizontal bar chart
plt.barh(categories,values,color="skyblue")
plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Qty")
plt.show()
```

Pie Charts -> circular chart divided into slices to show %age of the total

```python
categories=["Freshman","Sophomores","Juniors","Seniors"]
values=np.array([300,250,270,225])
shades=["red","yellow","blue","green"]
plt.pie(values,labels=categories,
        autopct="%1.1f%%", #"%1f" for only floating number but %1f%% for % sign at the end
        colors=shades,
        explode=[0,0,0,0.1],#jis part ko alag krna h jese 0.1,0,0,0 to first alag ho rha h
        shadow=True,
        startangle=180
        )
plt.title("Delhi Technological University")
plt.show()
```

Scatter Graphs -> shows the relationship b/w 2 vars,it helps to identify a correlation(+,-,None),eg: study hrs vs test scores

```python
x=np.array([0,1,1,2,3,4,5,6,7,7,8])#Hours studied
y=np.array([55,60,65,62,68,70,75,78,82,85,87])#grades
plt.xlabel("Hours Studies")
plt.ylabel("Grades")
plt.title("Test Scores")
plt.scatter(x,y,color="skyblue",
            alpha=0.5,#transparency
            s=200#size of dots
            )
plt.show()
```

```python
x1=np.array([0,1,1,2,3,4,5,6,7,7,8])#Hours studied
y1=np.array([55,60,65,62,68,70,75,78,82,85,87])#grades
x2=np.array([0,1,1,2,3,4,5,6,7,7,8])#Hours studied
y2=np.array([50,58,65,70,72,78,83,88,92,95,97])#grades
plt.xlabel("Hours Studies")
plt.ylabel("Grades")
plt.title("Test Scores")
plt.scatter(x1,y1,color="skyblue",
            alpha=0.5,#transparency
            s=200,#size of dots
            label="Class A")
plt.scatter(x2,y2,color="red",
            alpha=0.5,#transparency
            s=200,#size of dots
            label="Class B")
plt.legend()
plt.show()
```

Histograms -> visual representation of distribution of quantitative data

```python
scores=np.random.normal(loc=80,scale=10,size=100)
scores=np.clip(scores,0,100)
plt.hist(scores,bins=10,
         color="lightgreen",
         edgecolor="black")
plt.title("Exam Scores")
plt.xlabel("Scores")
plt.ylabel("No. of Students")
plt.show()
```

Subplots -> figure:entire canvas Ax:a single plot

```python
x=np.array([1,2,3,4,5])
figure,axes=plt.subplots(2,2)#subplots(rows,columns)
#axes is actually a numpy array 
axes[0,0].plot(x,x*2,color="red")
axes[0,0].set_title("x*2")

axes[0,1].plot(x,x**2,color="blue")
axes[0,1].set_title("x**2")

axes[1,0].plot(x,x**3,color="green")
axes[1,0].set_title("x**3")

axes[1,1].plot(x,x**4,color="purple")
axes[1,1].set_title("x**4")

plt.tight_layout()
plt.show()
```
